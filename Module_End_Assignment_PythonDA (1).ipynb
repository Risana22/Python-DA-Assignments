### Task 5 — Statistical Analysis

Compute descriptive stats for likes, comments, shares, watch_time, engagement_rate, followers:

-->Mean, median, mode

-->Standard deviation, variance

-->Percentiles

-->(Optional) Skewness and kurtosis



**Descriptive statistics like mean,median,mode are the measures of central tendency which gives the typical values of the data like average,middle value and most frequently occuring value currespondingly.variance and standard deviation are the measures of dispersion, which  shows the spread of the data, i.e; when increase in standard deviation tells that the data is widely spread. Finally percentiles is the measure of possition which devides the data into 4quarters and each quartiles gives the lower quartile,middle quartile,upper boundary of the middle quartile,upper quartile currespondingly.** .

```python
from numpy._core.fromnumeric import var
print("Descriptive stats for comments field")
mean1=df["comments"].mean().round()
print("Average of  comments field: ",mean1)
median1=df["comments"].median()
print("Median of  comments field: ",median1)
mode1=df["comments"].mode()[0]
print("Mode of  comments field: ",mode1)
std1=df["comments"].std()
print("Standard deviation of  comments field: ",std1)
var1=df["comments"].var()
print("Variance of  comments field: ",var1)
percentiles1=df["comments"].quantile([0.25,0.50,0.75,0.90,0.95])
print("Percentiles of  comments field: ",percentiles1)
```

**From the descriptive stats of comments field we can conclude that it is a left skewed data,since mean< median=mode. i.e; it shows the slightly negative skewed distribution that is most of the posts got high comments but some of them got very low comments that turned the mean lower than median.The variation can be understandable from the difference between values in 75th and 90th percentiles.From the std and var it is clear that the data is widely spread,ie; the comments column values are varied like high and low.**

### Plotly (Interactive)
Interactive line chart/bar chart/bubble/scatter chart



```python
import plotly.express as px
df["posted_day"]=df["posted_at"].dt.date
daily_engagement=df.groupby("posted_day")["engagement_rate"].sum()
fig=px.line(df,x=daily_engagement.index,y=daily_engagement.values,title="Daily Engagement Trends",
            color_discrete_sequence=px.colors.qualitative.Set3)
fig.update_layout(xaxis_title="Posted_date")
fig.update_layout(yaxis_title="Engagaement_rate")
fig.update_traces( mode="lines+markers",marker_size=8,marker_symbol="star",line_dash="dashdot")
fig.update_layout(title_font_size=20,template="plotly_dark")
fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=False)

fig.show()
```

**Here a daily engagement trend is prepared in a line plot using plotly(interactive and dynamic). dt,date keyword is used for extracting day from the posted date. And that is used in group by so that we will get daily engagement rate.
Extra feature in the line like,axis titles,template,fint size are given using .update_layout keyword, line style mode is set using update_traces, removed grids using update.axes.**

**Analysis on visual:-**

**From the above line plot it is clear that there is no particular trend or pattern in the engagement rate in daily basis.There are ups and downs in the engagement rate, that shows the inconsistency of the engagement of poepleand the  the social media, So there has to be many changes in setting a trend in social media, great imrovement in the recommenadations for the people,improving encouragement for the impression marking, improvement in the content performance,etc**
