```python
Pandas Series:
1. Creating Pandas Series:
● Create a Pandas Series named marks with the following data:
➢ Marks: 95, 92, 89, 85, 80
➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'
```

```python
2. Indexing and Slicing:
● Use the integer index position to access the mark of the 1st rank student.
```

```python
print("Mark of the 1st rank student: ",marks.iloc[0])
```

```python
Here iloc is used for the assessing of 1st rank mark, since iloc is an  index based positioning method, index is used here.
```

```python
● Use the loc accessor to retrieve the marks of the top 3 ranks by
specifying their index labels.
```

```python
print("Marks of the top 3 ranks: ", marks.loc["Rank1":"Rank3"])
```

```python
Since loc is a label based accessor the labels are used with loc for slicing.
```

```python
● Use the iloc accessor to retrieve the mark of the 3rd rank student.
```

```python
print("Mark of the 3rd rank student: ",marks.iloc[2])
```

```python
Since iloc is used here so index is used for retreival.
```

```python
● Apply a boolean mask to filter the Series and retrieve the ranks where the
marks are greater than 90.
```

```python
print("The ranks have higher marks than 90: ", marks[(marks>90)])
```

```python
We need to filter things with certain conditions so we applied boolean filtering here.
```

```python
marks["Rank1"]=100
marks
```

```python
Here marks is modified in the specific index position by using label index of the marks.
```

```python
● Remove the entry corresponding to the last rank from the Series.
```

```python
marks=marks.drop(["Rank5"])
```

```python
marks
```
