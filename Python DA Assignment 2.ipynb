```python
Finally the categorical columns contains null values are imputed with first mode value from the mode of each columns. For that
here fillna() keyword is used, that can impute null values with the value given by us.
```

```python
df["payment"]=df["payment"].fillna(df["payment"].mode()[0])
df["pickup_zone"]=df["pickup_zone"].fillna(df["pickup_zone"].mode()[0])
df["dropoff_zone"]=df["dropoff_zone"].fillna(df["dropoff_zone"].mode()[0])
df["pickup_borough"]=df["pickup_borough"].fillna(df["pickup_borough"].mode()[0])
df["dropoff_borough"]=df["dropoff_borough"].fillna(df["dropoff_borough"].mode()[0])
```

```python
3) Visualizations using Matplotlib/Pandas Plot:
★ Line Chart
Plot a line chart to visualize the fare over time, using the pickup timestamp as the
x-axis and fare as the y-axis. Ensure the pickup column is converted to a datetime
format before plotting.
```

```python
df["pickup"]=pd.to_datetime(df["pickup"],errors="coerce")
```

```python
Alraedy the column pickup is in datetime dtype, but for the assurance of all the rows the above code is used to change it into datetime datatype.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df["pickupday"]=df["pickup"].dt.date
daily_fare=df.groupby("pickupday")["fare"].sum()
plt.plot(daily_fare.index,daily_fare.values,data=df,color="lime",marker="D",markersize=3)
plt.title("Line chart showing daily total fare over time")
plt.xlabel("Pickup time stamp")
plt.ylabel("Fare")
plt.show()
```

```python
Here both pandas and matplotlib is imported for the visualization. From the pickup date stamp, date is extracted for better visualization, 
since the pickup date data may vary the visualization very crowded and that will not provide a better analysis. And group by is used for 
grouping pickup day accoring to the fare.Grouping is done using .sum() method.
              Then in the plot on the x-axis day time stamp is taken and on the y-axis fare is taken.For making the visul more meaningful and 
understandable extra features like marker,markersize,color are used when defining lineplot.
Analysis on visual:
The data shows the sharp hike in the fare during starting days, then it leads to the small up and down variations through day by day, 
the analysis have a hidden idea that everyday's trip will definitely influence the total fare of that day. That is if the total trip per day 
increases the total fare is also increases, that means the reason for the variation of total fare in each day is due to the number of trips
in each day.

```

```python
★ Box Plot
Plot a box plot to visualize the distribution of tip amounts for each pickup_borough.
Use pickup_borough as the categorical axis and tip as the numeric axis.
```

```python
boroughs=df["pickup_borough"].unique()
tip_data=[df[df["pickup_borough"]==borough]["tip"] for borough in boroughs]
plt.boxplot(tip_data,label=boroughs,patch_artist=True)
plt.xlabel("Pickup_borough")
plt.ylabel("Tip")
plt.title("Boxplot showing distribution of tip in each pickup_borough")
plt.show()
```

```python
.unique() keyword is used here to take all pickup_borugh as the boxplot to separately identify their outliers.
Borough is used as a variable to store each pickup_borough, and for loop is used here to iterate over each borough.
In the tip_data variable the pickup_boprough individualy categorized along with tip for get the status of tip in each borough region.
As an extra feature patch artist keyword is used here to color the boxplot.
Analysis on visual:
    From the boxplot of each pickup_borough it is clear that the status of tip shows infinite outliers than the upper fence of the data in the first
pickup_borough region, the second region show few outliers but the median Q2 is similar with the Q1, the third region shows few outliers, but there
is no definite Q1,Q2,Q3 and range of tip that the most of the passenegers in thet area couldn't prefer to give tip, the final borough shows many
outliers and median is similar to Q1, so from this visual it is clear that the amount of tip is completely based on the passengers mindset,
that shows the infinite number of outliers in the first and last regions.and in the 2nd region the passengers gives more amount as tip as 
comparing to other regions.So it is clear that the tip amount is completely depend on the passenger mindset but anyhow the mindset of some regions
like region1 and 4 shows same that's why their boxplot is like this.
```
